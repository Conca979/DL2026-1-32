```python
# Cell 2: Paths (hardcoded for Kaggle)

import os
user = os.listdir("/kaggle/input/datasets")[0]
path = f"/kaggle/input/datasets/{user}"

print(user)
print(os.listdir(path))

from pathlib import Path
import shutil

SOURCE_DIR = Path(f"{path}/nct-crc-he-100k-nonorm/NCT-CRC-HE-100K-NONORM")
TARGET_DIR = Path(f"{path}/crc-val-he-7k/CRC-VAL-HE-7K")
WORKING    = Path("/kaggle/working")

# Copy run_experiments.py from the attached code dataset
script_dst = WORKING / "run_experiments.py"
shutil.copy2(
    next(Path(f"{path}/histo-robust-code").rglob("run_experiments.py")),
    script_dst
)

print("Source:", SOURCE_DIR, "| exists:", SOURCE_DIR.exists())
print("Target:", TARGET_DIR, "| exists:", TARGET_DIR.exists())
print("Script:", script_dst, "| exists:", script_dst.exists())
```

```python
# Cell 3: Execute All Ablation Experiments
import subprocess, sys

if not script_dst.exists():
    raise FileNotFoundError(f"Cannot find {script_dst}. Ensure histo-robust-code dataset is attached.")

results_dir = WORKING / "results"
results_dir.mkdir(parents=True, exist_ok=True)

cmd = [
    sys.executable,
    str(script_dst),
    "--data-root", str(SOURCE_DIR),
    "--target-root", str(TARGET_DIR),
    "--out-dir", str(results_dir),
    "--subset", "25000",
    "--epochs", "8",
    "--batch-size", "64",
]

print("Starting ablation pipeline:", " ".join(cmd))
proc = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
for line in proc.stdout:
    print(line, end="", flush=True)
proc.wait()

if proc.returncode != 0:
    raise RuntimeError(f"Ablation suite exited with error code {proc.returncode}")
```
